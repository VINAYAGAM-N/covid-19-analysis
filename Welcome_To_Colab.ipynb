```
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# Load data
df = pd.read_csv("agri.csv")

# Central Tendency
mean_price = df['Modal_x0020_Price'].mean()
median_price = df['Modal_x0020_Price'].median()
mode_price = df['Modal_x0020_Price'].mode()[0]

# Dispersion
range_price = df['Modal_x0020_Price'].max() - df['Modal_x0020_Price'].min()
std_dev = df['Modal_x0020_Price'].std()
iqr = df['Modal_x0020_Price'].quantile(0.75) - df['Modal_x0020_Price'].quantile(0.25)

# Shape
skewness = skew(df['Modal_x0020_Price'])
kurt = kurtosis(df['Modal_x0020_Price'])

# Print Descriptive Statistics with Labels
print("Measures of Central Tendency:")
print(f"Mean Modal Price: {mean_price}")
print(f"Median Modal Price: {median_price}")
print(f"Mode Modal Price: {mode_price}")
print("Measures of Dispersion (Spread):")
print(f"Range of Modal Prices: {range_price}")
print(f"Standard Deviation: {std_dev}")
print(f"IQR: {iqr}")
print("Shape of the Distribution:")
print(f"Skewness: {skewness}")
print(f"Kurtosis: {kurt}")

# Plot Histogram to Visualize Distribution
plt.figure(figsize=(10, 6))
plt.hist(df['Modal_x0020_Price'], bins=15, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(mean_price, color='red', linestyle='dashed', linewidth=1.5, label=f'Mean: {mean_price:.2f}')
plt.axvline(median_price, color='green', linestyle='dashed', linewidth=1.5, label=f'Median: {median_price:.2f}')
plt.axvline(mode_price, color='orange', linestyle='dashed', linewidth=1.5, label=f'Mode: {mode_price}')

# Add Skewness and Kurtosis Annotations
plt.title("Modal Price Distribution", fontsize=14)
plt.xlabel("Modal Price", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.legend()
plt.show()
```

```
import pandas as pd
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
from imblearn.combine import SMOTEENN
from collections import Counter
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("agri.csv")

# Add a categorical target variable for demonstration
# Ensure the bins are appropriate for your data to avoid empty categories
df['Price Category'] = pd.cut(df['Modal_x0020_Price'], bins=[df['Modal_x0020_Price'].min()-1,
                                                            df['Modal_x0020_Price'].quantile(0.33),
                                                            df['Modal_x0020_Price'].quantile(0.66),
                                                            df['Modal_x0020_Price'].max()+1],
                               labels=['Low', 'Medium', 'High'])
# Instead of dropping NaNs, consider imputing or using a strategy to handle them
# For this example, we'll fill NaNs with the most frequent category
df['Price Category'] = df['Price Category'].fillna(df['Price Category'].mode()[0])

# Features and Target
X = df[['Min_x0020_Price', 'Max_x0020_Price']]
y = df['Price Category']

# Check class distribution
print("Original class distribution:", Counter(y))

# Oversampling using SMOTE
smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)
print("After SMOTE:", Counter(y_smote))

# Undersampling
undersampler = RandomUnderSampler(random_state=42)
X_under, y_under = undersampler.fit_resample(X, y)
print("After Undersampling:", Counter(y_under))

# Combination SMOTE and Undersampling using SMOTEENN
smoteenn = SMOTEENN(random_state=42)
X_smoteenn, y_smoteenn = smoteenn.fit_resample(X, y)
print("After SMOTEENN:", Counter(y_smoteenn))

# Visualization of the Resampled Data
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

axes[0].bar(Counter(y_smote).keys(), Counter(y_smote).values(), color='skyblue')
axes[0].set_title("Oversampling (SMOTE)")
axes[1].bar(Counter(y_under).keys(), Counter(y_under).values(), color='lightgreen')
axes[1].set_title("Undersampling")
axes[2].bar(Counter(y_smoteenn).keys(), Counter(y_smoteenn).values(), color='salmon')
axes[2].set_title("Combination (SMOTEENN)")

for ax in axes:
    ax.set_xlabel("Class")
    ax.set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```
